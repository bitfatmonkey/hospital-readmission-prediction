# 02 · Cohort, clinical features and leakage controls

Prediction is at **discharge**, not admission. Death/hospice and ongoing admissions are excluded before splitting. No rows are resampled and no synthetic observations are generated.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "hospital_readmission").is_dir())
def report(name):
    return json.loads((ROOT / "reports" / name).read_text())
def figure(name):
    display(Image(filename=str(ROOT / "figures" / name)))
from hospital_readmission.data import load_diabetes
from hospital_readmission.cohort import make_cohort
from hospital_readmission.config import ExperimentConfig
from hospital_readmission.splitting import split_patients
dataset = load_diabetes()
cohort, cohort_audit = make_cohort(dataset)
splits, split_audit = split_patients(cohort, ExperimentConfig())
training = cohort.iloc[splits["train"]]
print("Real ZIP loaded offline:", dataset.data.shape, "eligible:", cohort.shape)
```

```python
display(pd.DataFrame(cohort_audit["exclusions"]))
print("Eligible encounters:", len(cohort), "patients:", cohort.patient_nbr.nunique())
```

## Prespecified feature engineering
Derive age midpoint, broad diagnosis categories, prior-visit totals, medication indicators and test-measurement flags. These rules are stateless. Discharge code IDs are categorical. Drop identifiers, the target, race, gender and payer. Weight is removed only if its missingness exceeds 95% on the current training fold.

```python
from hospital_readmission.features import ClinicalFeatureEngineer
engineered = ClinicalFeatureEngineer().fit_transform(training)
display(pd.DataFrame({"dtype":engineered.dtypes.astype(str), "missing_fraction":engineered.isna().mean()}))
print("Engineered columns:", engineered.columns.tolist())
```

## Patient-level holdouts and cross-validation
Five-fold stratified group splits reserve approximately 20% for test and a further 20% of development for threshold validation. Three group-aware CV folds within training tune models. The same patients never appear on both sides of any split.

```python
manifest = report("split_manifest.json")
display(pd.DataFrame({k:manifest[k] for k in ["train","validation","test"]}).T)
display(pd.DataFrame(manifest["training_cv"]))
assert all(v == 0 for v in manifest["pairwise_patient_overlap"].values())
```

## Training-fitted pipeline
Each CV candidate owns its feature engineering, missingness rule, median imputation and one-hot category dictionary. Logistic regression additionally scales numeric features. Class weights are candidates inside training CV, rather than oversampling before splitting. Validation and test transformations never refit these steps.

```python
import joblib
artifact = joblib.load(ROOT / "artifacts/selected_model.joblib")  # locally trained/trusted only
display(artifact["pipeline"])
prep = report("preprocessing.json")
print("Removed for training missingness:", prep["dropped_for_missingness"])
print("Encoded features:", len(prep["encoded_feature_names"]))
print("Excluded predictors:", prep["excluded_predictors"])
```

## Limitations of the prediction snapshot
The encounter summaries are assumed available at discharge. Exact coding/test-result availability is not timestamped, so operational real-time availability is unproven. Admission/discharge dates are absent, preventing a true temporal validation or an overlap-in-time audit across distinct patients. Transfers and incomplete readmission capture remain label limitations.
